```python
episodes_with_id = pd.read_csv("../data/episodes_with_id.csv")
```

```python
import os

# Initialize an empty DataFrame
filtered_df = pd.DataFrame()

# Get the list of episode IDs from episodes_with_id
valid_episode_ids = set(episodes_with_id['episode_id'])

# Directory containing the CSV files
directory = '../data/episodes/speaker_turns'

# Loop through the files in the directory
for file_name in os.listdir(directory):
    if file_name.startswith('episode_') and file_name.endswith('.csv'):
        # Extract the episode_id from the file name
        episode_id = int(file_name.split('_')[1].split('.')[0])
        
        # Check if the episode_id is in the valid_episode_ids
        if episode_id in valid_episode_ids:
            # Read the CSV file and append it to the filtered_df
            file_path = os.path.join(directory, file_name)
            temp_df = pd.read_csv(file_path)
            filtered_df = pd.concat([filtered_df, temp_df], ignore_index=True)

# Display the resulting DataFrame
filtered_df
```

```python
racial_justice_count = filtered_df[filtered_df['racialJustice'] == 1].shape[0]
binary_action_count = filtered_df[filtered_df['binaryAction'] == 0].shape[0]
combined_count = filtered_df[(filtered_df['racialJustice'] == 1) & (filtered_df['binaryAction'] == 0)].shape[0]
racial_justice_count, binary_action_count, combined_count
```
