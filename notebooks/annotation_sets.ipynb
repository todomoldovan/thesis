### Three sets of annotations

```python
pd.set_option('display.max_colwidth', None)
```

1. Racial justice

```python
import pandas as pd

# change the annotation file paths when received
file_paths = [
    '../data/annotation_set_race_davide.csv',
    '../data/annotation_set_race_arianna.csv',
    '../data/annotation_set_race_theodora.csv'
]
labels_path = '../data/annotation_set_race_labels.csv'

dfs = [pd.read_csv(file_path) for file_path in file_paths]
labels_df = pd.read_csv(labels_path)

# Identify the non-NaN columns for each annotator
annotator_columns = []
for df in dfs:
    for col in ['annotator1', 'annotator2', 'annotator3']:
        if df[col].notna().any():
            annotator_columns.append(df[col])
            break

# Add the 'note' column from Arianna's annotation set
arianna_notes = dfs[1]['note'] if 'note' in dfs[1].columns else None

# Create a new DataFrame with the sentence column and the three annotator columns
race_annotations = pd.DataFrame({
    'sentence': dfs[0]['sentence'],
    'davide': annotator_columns[0],
    'arianna': annotator_columns[1],
    'theodora': annotator_columns[2],
    'arianna_note': arianna_notes
})

# Function to compute majority vote
def majority_vote(row):
    return row.mode().iloc[0] if not row.mode().empty else None

# Apply majority vote function
race_annotations['majority'] = race_annotations[['davide', 'arianna', 'theodora']].apply(majority_vote, axis=1)

# Merge with racialJustice column from labels_df
race_annotations = race_annotations.merge(labels_df[['sentence', 'racialJustice']], on='sentence', how='left')

# Display the result
race_annotations.head()
```

```python
race_annotations[['sentence', 'majority']].to_csv('../data/annotation_set_race_majority.csv', index=False)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix

# Compute accuracy
accuracy = accuracy_score(race_annotations['racialJustice'], race_annotations['majority'])

# Compute confusion matrix
tn, fp, fn, tp = confusion_matrix(race_annotations['racialJustice'], race_annotations['majority']).ravel()

# Print results
print(f'Accuracy: {accuracy:.4f}')
print(f'True Positives (TP): {tp}')
print(f'True Negatives (TN): {tn}')
print(f'False Positives (FP): {fp}')
print(f'False Negatives (FN): {fn}')
```

```python
import krippendorff
import numpy as np

# Extract the relevant columns
data = race_annotations[['davide', 'arianna', 'theodora']].to_numpy()
data = np.transpose(data)

# Calculate Krippendorff's alpha
alpha = krippendorff.alpha(reliability_data=data, level_of_measurement='nominal')

print(f"Krippendorff's alpha: {alpha:.4f}")
```

2. Collective action

```python
import pandas as pd

# change the annotation file paths when received
file_paths = [
    '../data/annotation_set_action_davide.csv',
    '../data/annotation_set_action_arianna.csv',
    '../data/annotation_set_action_theodora.csv'
]
labels_path = '../data/annotation_set_action_labels.csv'

dfs = [pd.read_csv(file_path) for file_path in file_paths]
labels_df = pd.read_csv(labels_path)

# Identify the non-NaN columns for each annotator
annotator_columns = []
for df in dfs:
    for col in ['annotator1', 'annotator2', 'annotator3']:
        if df[col].notna().any():
            annotator_columns.append(df[col])
            break

arianna_notes = dfs[1]['note'] if 'note' in dfs[1].columns else None

# Create a new DataFrame with the sentence column and the three annotator columns
action_annotations = pd.DataFrame({
    'sentence': dfs[0]['sentence'],
    'davide': annotator_columns[0],
    'arianna': annotator_columns[1],
    'theodora': annotator_columns[2],
    'arianna_note': arianna_notes
})

# Function to compute majority vote
def majority_vote(row):
    return row.mode().iloc[0] if not row.mode().empty else None

# Apply majority vote function
action_annotations['majority'] = action_annotations[['davide', 'arianna', 'theodora']].apply(majority_vote, axis=1)
action_annotations['majority'] = action_annotations['majority'].astype(int)

# Directly copy the 'collectiveAction' column from labels_df to action_annotations
action_annotations['collectiveAction'] = labels_df['collectiveAction']

# Display the result
action_annotations.head()
```

```python
action_annotations[['sentence', 'majority']].to_csv('../data/annotation_set_action_majority.csv', index=False)
```

```python
print(action_annotations[['davide', 'arianna', 'theodora']].dtypes)
print(action_annotations[['davide', 'arianna', 'theodora']].apply(pd.unique))
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix

action_annotations['collectiveAction'] = 1 - action_annotations['collectiveAction']

# Compute accuracy
accuracy = accuracy_score(action_annotations['collectiveAction'], action_annotations['majority'])

# Compute confusion matrix
tn, fp, fn, tp = confusion_matrix(action_annotations['collectiveAction'], action_annotations['majority']).ravel()

# Print results
print(f'Accuracy: {accuracy:.4f}')
print(f'True Positives (TP): {tp}')
print(f'True Negatives (TN): {tn}')
print(f'False Positives (FP): {fp}')
print(f'False Negatives (FN): {fn}')
```

```python
import krippendorff
import numpy as np
import pandas as pd

valid_values = {0, 1}

# Convert all columns to numeric (coerce errors to NaN)
action_annotations[['davide', 'arianna', 'theodora']] = action_annotations[['davide', 'arianna', 'theodora']].apply(pd.to_numeric, errors='coerce')

# Drop rows that contain invalid values (anything other than 0 or 1)
filtered_data = action_annotations.dropna(subset=['davide', 'arianna', 'theodora'])
filtered_data = filtered_data[
    filtered_data[['davide', 'arianna', 'theodora']].applymap(lambda x: x in valid_values).all(axis=1)
]

data = filtered_data[['davide', 'arianna', 'theodora']].to_numpy(dtype=int).T
alpha = krippendorff.alpha(reliability_data=data, level_of_measurement='nominal')
print(f"Krippendorff's alpha: {alpha:.4f}")
```
