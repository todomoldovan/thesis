181k episodes now have all newSpeakers in mainEpSpeakers

```python
conn = sqlite3.connect('../data/data.db')
cursor = conn.cursor()

query = f'''
WITH RECURSIVE split_speakers(mp3url, mainEpSpeaker, rest) AS (
    -- Base case: Extract first speaker
    SELECT 
        mp3url,
        TRIM(SUBSTR(mainEpSpeakers, 1, INSTR(mainEpSpeakers || ',', ',') - 1)) AS mainEpSpeaker,
        SUBSTR(mainEpSpeakers, INSTR(mainEpSpeakers || ',', ',') + 1) AS rest
    FROM speaker_turns
    WHERE mainEpSpeakers IS NOT NULL

    UNION ALL

    -- Recursive case: Extract next speaker
    SELECT 
        mp3url,
        TRIM(SUBSTR(rest, 1, INSTR(rest || ',', ',') - 1)),
        SUBSTR(rest, INSTR(rest || ',', ',') + 1)
    FROM split_speakers
    WHERE rest <> '' 
    AND INSTR(rest, ',') > 0  -- Only continue if there's more to split
    OR rest != ''             -- When no comma is left, handle the final speaker
),

-- CTE to get all speakers per episode
main_speakers AS (
    SELECT DISTINCT mp3url, mainEpSpeaker 
    FROM split_speakers
),

-- CTE to get all newSpeakers per episode
new_speakers AS (
    SELECT DISTINCT mp3url, newSpeaker
    FROM speaker_turns
)

SELECT COUNT(DISTINCT st.mp3url) AS num_episodes
FROM speaker_turns st
JOIN new_speakers ns ON st.mp3url = ns.mp3url
LEFT JOIN main_speakers ms ON st.mp3url = ms.mp3url AND ns.newSpeaker = ms.mainEpSpeaker
GROUP BY st.mp3url
HAVING SUM(CASE WHEN ms.mainEpSpeaker IS NULL THEN 1 ELSE 0 END) = 0;
'''

cursor.execute(query)

rows = cursor.fetchall()

column_names = [description[0] for description in cursor.description]

print(tabulate(rows, headers=column_names, tablefmt="pretty"))

# Close the cursor and the connection
cursor.close()
conn.close()
```

Make new column newMainEpSpeakers for improved speaker turns

```python
import sqlite3

# Connect to your SQLite database
conn = sqlite3.connect('../data/data.db')
cursor = conn.cursor()

# Step 1: Add the new column 'newMainEpSpeakers' to the table
cursor.execute('ALTER TABLE speaker_turns ADD COLUMN newMainEpSpeakers TEXT')

# Step 2: Select the unique mp3urls and associated speakers
cursor.execute("""
    SELECT mp3url, newSpeaker
    FROM speaker_turns
""")
rows = cursor.fetchall()

# Dictionary to store unique speakers for each mp3url
episode_speakers = {}

# Process the rows to collect speakers by mp3url
for mp3url, newSpeaker in rows:
    if mp3url not in episode_speakers:
        episode_speakers[mp3url] = set()

    # Split the speakers and add them to the set to ensure uniqueness
    speakers = newSpeaker.split(',') if newSpeaker else []
    for speaker in speakers:
        episode_speakers[mp3url].add(speaker.strip())  # Strip whitespace and add unique speakers

# Step 3: Update the 'newMainEpSpeakers' column with the unique speakers for each mp3url
for mp3url, speakers in episode_speakers.items():
    # Convert the set of unique speakers to a string format (list)
    unique_speakers_str = str(sorted(list(speakers)))  # Sort for consistency
    cursor.execute("""
        UPDATE speaker_turns
        SET newMainEpSpeakers = ?
        WHERE mp3url = ?
    """, (unique_speakers_str, mp3url))

# Commit the changes
conn.commit()

# Close the connection
conn.close()

print("The 'newMainEpSpeakers' column has been updated.")
```

Copy column numNewMainEpSpeakers from speaker_turns to podcast_episodes

```python
import sqlite3

# Connect to the database
conn = sqlite3.connect('../data/data.db')
cursor = conn.cursor()

# Step 1: Check and add column if it doesn't exist
cursor.execute("PRAGMA table_info(podcast_episodes);")
columns = [col[1] for col in cursor.fetchall()]
if "numNewMainEpSpeakers" not in columns:
    print("Adding column numNewMainEpSpeakers...")
    cursor.execute("ALTER TABLE podcast_episodes ADD COLUMN numNewMainEpSpeakers TEXT;")
    conn.commit()

# Step 3: Batch processing to avoid massive single-query execution
batch_size = 100_000
offset = 0

# Check total rows for progress tracking
cursor.execute("SELECT COUNT(*) FROM podcast_episodes;")
total_rows = cursor.fetchone()[0]
print(f"Total rows to process: {total_rows}")

while offset < total_rows:
    print(f"Processing batch starting at row {offset}...")
    
    # Update a batch of rows
    cursor.execute(f"""
        UPDATE podcast_episodes
        SET numNewMainEpSpeakers = (
            SELECT pe.numNewMainEpSpeakers
            FROM speaker_turns pe
            WHERE podcast_episodes.mp3url = pe.mp3url
        )
        WHERE rowid IN (
            SELECT rowid FROM podcast_episodes
            LIMIT {batch_size} OFFSET {offset}
        );
    """)
    
    conn.commit()
    offset += batch_size

print("Update process completed.")
# Close the connection
conn.close()
```
